**Note:** This notebook does not perform dataset processing,

```python
#Check Python

import sys
import platform

print("Python version:", sys.version)
print("Platform:", platform.platform())
print("Arcitecture:",platform.machine())
```

```python
# Check PyTorch
import torch
print("Torch version:", torch.__version__)
print("Torch CUDA available:", torch.cuda.is_available())
```

```python
# Check GPU
if torch.cuda.is_available():
    print("GPU Name:", torch.cuda.get_device_name(0))
    print("GPU Count:", torch.cuda.device_count())
    print("CUDA Version:", torch.version.cuda)
else:
    print("Warning: No GPU available.")
```

```python
# Check GPU memory
if torch.cuda.is_available():
    gpu = torch.cuda.get_device_properties(0)
    print("GPU:",gpu.name)
    print("Total GPU Memory:", gpu.total_memory / (1024 ** 3), "GB")
```

```python
# Actual test the GPU

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Selected device:", device)
```

```python
x = torch.randn(1000, 1000).to(device)
y = torch.matmul(x, x)

print("Tensor device:", y.device)
print("GPU computation successful!")
```

```python
import os

print("Google Drive contents:")
print(os.listdir('/content/drive/MyDrive'))
```

```python
PROJECT_DIR = "/content/drive/MyDrive/DeepGuard"

folders = [
    "checkpoints",
    "results",
    "samples"
]

for folder in folders:
    path = os.path.join(PROJECT_DIR, folder)
    os.makedirs(path, exist_ok=True)

print("DeepGuard folders created successfully!")
```

```python
for root, dirs, files in os.walk(PROJECT_DIR):
    level = root.replace(PROJECT_DIR, "").count(os.sep)
    indent = "    " * level

    print(f"{indent}{os.path.basename(root)}/")
```

```python
# Test that Colab can write to Drive

test_file = os.path.join(
    PROJECT_DIR,
    "results",
    "environment_test.txt"
)

with open(test_file, "w") as f:
    f.write("DeepGuard environment test successful.")

print("Test file created successfully!")
```

```python
with open(test_file, "r") as f:
    print(f.read())
```

```python
# Check Temp Storage

import shutil

total, used, free = shutil.disk_usage("/content")

print("Colab storage:")
print("Total:", round(total / (1024**3), 2), "GB")
print("Used :", round(used / (1024**3), 2), "GB")
print("Free :", round(free / (1024**3), 2), "GB")
```
